Write a script that would work as Automated teller machine (ATM). It would have like Deposit, check balance, withdraw. As the owner of my account, I have to put the initial money to be on the account through deposit, then if I check balance it would reflect, then if I withdraw it would debit the exact amount and the balance would reflect on the check balance, then if I withdraw to the extent that the initial amount as balance is finished, I can't withdraw again, but it will print something like "Insufficient fund", then I can use the deposit to put another amount.

```python

def atm_simulator():
  balance = 1000.00  # Initial account balance

  while True:
    print("\n--- ATM Menu ---")
    print("1. Check Balance")
    print("2. Deposit Money")
    print("3. Withdraw Money (Debit)")
    print("4. Exit")

    choice = input("Choose an option (1-4): ")

    if choice == "1":
      print(f"Current Balance: ${balance:.2f}")

    elif choice == "2":
      amount = float(input("Enter amount to deposit: "))
      if amount > 0:
        balance += amount
        print(f"Successfully deposited ${amount:.2f}.")
        print(f"New Balance: ${balance:.2f}")
      else:
        print("Invalid deposit amount.")

    elif choice == "3":
      amount = float(input("Enter amount to withdraw: "))
      if 0 < amount <= balance:
        balance -= amount
        print(f"Successfully debited ${amount:.2f}.")
        print(f"New Balance: ${balance:.2f}")
      elif amount > balance:
        print("Error: Insufficient funds!")
      else:
        print("Invalid withdrawal amount.")

    elif choice == "4":
      print("Thank you for using the ATM. Goodbye!")
      break
    else:
      print("Invalid choice. Please choose between 1 and 4.")


# Run the ATM
atm_simulator()
```
