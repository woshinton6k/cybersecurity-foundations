## Infinite-loop concept (safe explanation)

An infinite loop continues while its condition never becomes false. Running one directly in a notebook can make the kernel appear unresponsive. The next cell demonstrates repetition with a clear stopping condition instead of launching an endless loop.


```python
for iteration in range(1, 6):
    print(f"Demonstration iteration {iteration}")

print("The demonstration stopped safely after five iterations.")
```

```python
# 1. Define the function with two parameters (a, b)
def add_numbers(a, b):
    sum_result = a + b
    # Send the value back to where the function was called
    return sum_result

# 2. Use the function
# We store the returned value in a variable called 'total'
total = add_numbers(5, 10)

print("The calculation result is:", total)
```

```python
import re

# 1. Define the text and the word you are searching for
text = "I am currently learning Python programming language at Digital dreams. My name is Big-wosh."
search_word = "Big-wosh"

# 2. Use re.search(pattern, string)
match = re.search(search_word, text)

# 3. Check if a match was found
if match:
    print("Match found!")
    print("The word was found at:", match.start()) # Shows the starting position
else:
    print("Word not found.")
```

```python
import re

text = "The year 2026 has 12 months and 365 days, it also has 52 weeks."

# 1. Find all digits
# \d matches any digit (0-9)
# + means "one or more" (it keeps digits together)
digits = re.findall(r'\d+', text)

# 2. Find all words
# \w matches any alphanumeric character (letters, numbers, underscore)
# + means "one or more"
words = re.findall(r'\w+', text)

# [a-zA-Z] means: Match any letter from a to z, or A to Z.
# The '+' still means: keep them grouped together as one word.
words_only = re.findall(r'[a-zA-Z]+', text)

print("Digits found:", digits)
print("Words found:", words)
print("Words only:", words_only)
```

# **Question 2B**

```python
from getpass import getpass

def password_checker():
    """Demonstrate repeated password input and attempt counting."""
    demo_password = getpass("Set a temporary demonstration password: ")
    attempts = 0

    while True:
        user_input = getpass("Enter the demonstration password: ")
        attempts += 1

        if user_input == demo_password:
            print(f"Access granted for this demonstration after {attempts} attempt(s).")
            break
        print("Incorrect password. Please try again.")

# Educational demonstration only; not suitable for production authentication.
password_checker()
```
