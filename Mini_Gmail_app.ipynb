create a script or a program that would validate 1. email 2. password with a specific length not more than or less than 8 characters, there must be symbols, digits, word, word must start with an uppercase. 3. should have a Nigerian based phone number number format i.e +234, pass them in to sign in function. it will be sequential.


```python
import re

# Storage lists and dictionaries for our mini app

inbox = []
sent_box = []


def send_mail():
    print("\n--- Send Mail ---")
    recipient = input("To (Recipient Gmail): ").strip()

    if not re.match(r"^\w+@gmail\.com$", recipient):
        print("✗ Invalid recipient email format.")
        return

    subject = input("Subject: ").strip()
    body = input("Message body: ").strip()

    # Using an f-string template to format the outgoing email
    template = f"""
    ==================================
    FROM: my_account@gmail.com
    TO: {recipient}
    SUBJECT: {subject}
    ----------------------------------
    {body}
    ==================================
    """

    sent_box.append(template)
    print("✓ Mail sent successfully using template!")

def receive_mail():
    print("\n--- Simulate Receive Mail ---")
    sender = input("From (Sender Gmail): ").strip()

    if not re.match(r"^\w+@gmail\.com$", sender):
        print("✗ Invalid sender email format.")
        return

    subject = input("Subject: ").strip()
    body = input("Message body: ").strip()

    # Using an f-string template to format the incoming email
    template = f"""
    [INBOX MESSAGE]
    FROM: {sender}
    SUBJECT: {subject}
    ----------------------------------
    {body}
    ----------------------------------
    """

    inbox.append(template)
    print("✓ New mail received in your inbox!")

def view_inbox():
    print("\n--- INBOX ---")
    if not inbox:
        print("Your inbox is empty.")
        return
    for i, mail in enumerate(inbox, 1):
        print(f"Email #{i}:{mail}")

def view_sent():
    print("\n--- SENT BOX ---")
    if not sent_box:
        print("Your sent box is empty.")
        return
    for i, mail in enumerate(sent_box, 1):
        print(f"Sent Mail #{i}:{mail}")

def mini_gmail():
    while True:
        print("\n=== MINI GMAIL APP ===")
        print("1. Send Mail")
        print("2. Receive Mail")
        print("3. View Inbox")
        print("4. View Sent Box")
        print("5. Exit")

        choice = input("Choose an option (1-5): ").strip()


        if choice == "1":
            send_mail()
        elif choice == "2":
            receive_mail()
        elif choice == "3":
            view_inbox()
        elif choice == "4":
            view_sent()
        elif choice == "5":
            print("\nClosing Mini Gmail. Goodbye!")
            break
        else:
            print("✗ Invalid choice. Please pick between 1 and 6.")

# Run the Mini Gmail application
mini_gmail()
```
